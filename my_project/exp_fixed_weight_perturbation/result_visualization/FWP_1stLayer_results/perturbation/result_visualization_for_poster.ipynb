# Phase 3: Biologically Inspired Hidden Perturbations

Perturbations are applied to the **1st hidden layer output**.
Two model variants are compared:
- **SGD-delay** (solid, filled circles): SNN with learnable axonal delays.
- **SGD** (dashed, × markers): baseline SNN without delays.

Three dataset variants: **Whole** (blue), **Part** (green), **Norm** (red).

Two sweeps are shown side by side, both of which scramble spike timing at
(near-)fixed spike count: relocation is unbounded within the window, jitter is
local.

## 3A-B: Relocation and Jitter

- **Per-Spike Random Relocation** — a fraction $f$ of each hidden neuron's spikes is
  removed and re-inserted at randomly chosen empty time bins, so the per-neuron spike
  count is preserved exactly and only timing is destroyed. Sweep: $f$ = 0.0 → 1.0.
  These are the SHD results from the Phase 2 realistic experiments
  (`code/realistic/shd/`), read in place so the two notebooks never diverge.
- **Per-Spike Jitter** — independent Gaussian noise (σ ms) added to each spike time in
  the hidden layer output. Sweep: σ = 0 → 25 ms.

Protocol note: jitter models are **trained from scratch** under the perturbation,
whereas relocation comes from the **eval-only** protocol — for each
`(dataset, delay-mode)` a clean-trained checkpoint is evaluated across the sweep.
The figure below therefore mixes an eval-only panel (a) with a
trained-under-perturbation panel (b); compare across panels with that in mind.

Both panels share the y-axis and a single legend.

```python
import json
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import matplotlib as mpl

mpl.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 11,
    "axes.labelsize": 13,
    "axes.titlesize": 14,
    "legend.fontsize": 9.5,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "figure.dpi": 150,
})

# Notebook lives in result_visualization/FWP_1stLayer_results/perturbation/;
# sweep results are read in place from exp_fixed_weight_perturbation/code/.
CODE_PERTURBATION_DIR = pathlib.Path("../../../code/perturbation")
CODE_REALISTIC_DIR = pathlib.Path("../../../code/realistic")

COLORS = {"Whole": "#1f77b4", "Part": "#2ca02c", "Norm": "#d62728"}


def load_sweep(path: pathlib.Path) -> tuple[list[float], np.ndarray, np.ndarray]:
    """Return (sigma_values, means_pct, stds_pct) from a sweep results JSON."""
    with open(path) as fh:
        raw = json.load(fh)
    sigmas = sorted(int(k) for k in raw)
    means = np.array([raw[str(s)]["mean"] for s in sigmas]) * 100
    stds = np.array([raw[str(s)]["std"] for s in sigmas]) * 100
    return sigmas, means, stds
```

```python
RELOCATION_LOG = CODE_REALISTIC_DIR / "shd/log"
JITTER_LOG = CODE_PERTURBATION_DIR / "jitter/log"

VARIANTS = ("Whole", "Part", "Norm")


def load_sweep_float(path: pathlib.Path) -> tuple[list[float], np.ndarray, np.ndarray]:
    """Like load_sweep but treats result-keys as floats (for p_d / f values)."""
    with open(path) as fh:
        raw = json.load(fh)
    items = sorted(((float(k), k) for k in raw), key=lambda p: p[0])
    xs = [p[0] for p in items]
    means = np.array([raw[k]["mean"] for _, k in items]) * 100
    stds = np.array([raw[k]["std"] for _, k in items]) * 100
    return xs, means, stds


def load_group(
    log_dir: pathlib.Path,
    filename_template: str,
    loader,
) -> dict[tuple[str, bool], tuple[list, np.ndarray, np.ndarray]]:
    """Load one perturbation's six sweeps, keyed by (dataset_variant, uses_delay).

    Args:
        log_dir: Directory holding the sweep result JSONs.
        filename_template: Template with ``{variant}`` and ``{mode}`` fields.
        loader: Either ``load_sweep`` (integer keys) or ``load_sweep_float``.

    Returns:
        Mapping from (variant, uses_delay) to (x_values, means_pct, stds_pct).
    """
    sweeps = {}
    for variant in VARIANTS:
        for uses_delay in (True, False):
            filename = filename_template.format(
                variant=variant.lower(),
                mode="delay" if uses_delay else "nodelay",
            )
            sweeps[(variant, uses_delay)] = loader(log_dir / filename)
    return sweeps


def draw_sweep_panel(
    ax: plt.Axes,
    sweeps: dict,
    title: str,
    xlabel: str,
    xlim: tuple[float, float],
) -> None:
    """Draw one perturbation sweep panel onto *ax*.

    SGD-delay curves are solid with filled circles, SGD (no delay) curves are
    dashed with x markers; colour encodes the dataset variant.
    """
    for uses_delay, style, markersize, alpha in [
        (True, "o-", 5, 0.12),
        (False, "x--", 6, 0.08),
    ]:
        for variant in VARIANTS:
            xs, means, stds = sweeps[(variant, uses_delay)]
            ax.plot(xs, means, style, color=COLORS[variant],
                    markersize=markersize, linewidth=1.8)
            ax.fill_between(xs, means - stds, means + stds,
                            color=COLORS[variant], alpha=alpha)

    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_xlim(*xlim)
    ax.grid(True, linestyle=":", linewidth=0.7, alpha=0.7)


def build_legend_handles(solid_label: str, dashed_label: str) -> list[mlines.Line2D]:
    """Build the shared legend: two line styles in column 1, variants in column 2.

    Matplotlib fills legend columns top-to-bottom, so a blank third entry keeps
    the line styles and the dataset variants in separate columns when the legend
    is drawn with ``ncol=2``.

    Args:
        solid_label: Label for the solid / filled-circle style.
        dashed_label: Label for the dashed / x-marker style.

    Returns:
        Handles ready to pass to ``ax.legend(handles=...)``.
    """
    blank = mlines.Line2D([], [], linestyle="none", marker="none", label="")
    return [
        mlines.Line2D([], [], color="black", marker="o", linestyle="-",
                      markersize=5, linewidth=1.8, label=solid_label),
        mlines.Line2D([], [], color="black", marker="x", linestyle="--",
                      markersize=6, linewidth=1.8, label=dashed_label),
        blank,
    ] + [
        mlines.Line2D([], [], color=COLORS[variant], linewidth=2.5, label=variant)
        for variant in VARIANTS
    ]


def draw_grid_figure(
    panels: list[tuple[dict, str, str, tuple[float, float]]],
    suptitle: str = "",
    legend_panel: int = 1,
) -> plt.Figure:
    """Draw a 1x2 sweep figure with panel labels and one shared legend.

    Args:
        panels: Two ``(sweeps, title, xlabel, xlim)`` tuples, left then right.
        suptitle: Main title drawn above the row; omitted when empty.
        legend_panel: Index into *panels* of the panel carrying the legend.

    Returns:
        The completed figure.
    """
    fig, axes = plt.subplots(1, 2, figsize=(11, 4.5), sharey=True)

    for ax, panel_label, (sweeps, title, xlabel, xlim) in zip(
        axes, "ab", panels
    ):
        draw_sweep_panel(ax, sweeps, title, xlabel, xlim)
        ax.text(-0.06, 1.04, f"({panel_label})", transform=ax.transAxes,
                fontsize=14, fontweight="bold")

    axes[0].set_ylabel("Test Accuracy (%)")
    axes[0].set_ylim(0, None)
    axes[legend_panel].legend(
        handles=build_legend_handles("SGD-delay", "SGD"),
        loc="upper right", framealpha=0.9, ncol=2)

    if suptitle:
        fig.suptitle(suptitle, y=1.01, fontsize=15)
    fig.tight_layout()
    return fig


# ---------- Load sweep results ----------
# Relocation reuses the Phase 2 realistic SHD logs (see code/realistic/shd/).
relocation_sweeps = load_group(
    RELOCATION_LOG, "shd_{variant}_{mode}_hidden_perturbation_results.json",
    load_sweep_float)
jitter_sweeps = load_group(
    JITTER_LOG, "jitter_{variant}_{mode}_jitter_sweep_results.json", load_sweep)
```

```python
# ---------- Figure 1: relocation | jitter ----------
# Both manipulations scramble spike timing at (near-)fixed spike count:
# relocation is unbounded within the window, jitter is local.
fig = draw_grid_figure([
    (relocation_sweeps, "Random Relocation on 1st Hidden Layer",
     r"Relocation Fraction $f$", (-0.02, 1.02)),
    (jitter_sweeps, "Jitter Perturbation on 1st Hidden Layer",
     r"Jitter Std Dev $\sigma$ (ms)", (-0.5, 26)),
])
#fig.savefig("all_perturbations_results.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 3C: Combined 1st + 2nd Layer Comparison (2x2)

The four sweep panels are combined into a single 2x2 figure for direct comparison:
rows correspond to the perturbed layer (1st, 2nd), columns to the perturbation type
(relocation, jitter). Colour and line-style conventions match the panels above
(colour = dataset variant; solid/filled circles = SGD-delay; dashed/x = SGD).

Protocol note: relocation (both rows) and 2nd-layer jitter all use the eval-only
protocol (a clean-trained checkpoint evaluated across the sweep), whereas 1st-layer
jitter uses models **trained from scratch** under the perturbation — so the
top-right panel is not directly comparable to the other three on that axis.

```python
# ---------- Load 2nd-layer sweep results (reuses this notebook's log dirs) ----------
relocation_sweeps_2nd = load_group(
    RELOCATION_LOG,
    "shd_2ndLayer_{variant}_{mode}_hidden_perturbation_results.json",
    load_sweep_float)
jitter_sweeps_2nd = load_group(
    JITTER_LOG,
    "jitter_2ndLayer_{variant}_{mode}_evalOnly_jitter_sweep_results.json",
    load_sweep)


def draw_2x2_figure(
    panels: list[list[tuple[dict, str, str, tuple[float, float]]]],
    row_labels: list[str],
    suptitle: str = "",
    legend_ax: tuple[int, int] = (0, 1),
) -> plt.Figure:
    """Draw a 2x2 grid of sweep panels with row labels and one shared legend.

    Args:
        panels: 2x2 nested list of ``(sweeps, title, xlabel, xlim)`` tuples,
            indexed as ``panels[row][col]``.
        row_labels: Label prepended to the y-axis of each row (e.g. layer name).
        suptitle: Main title drawn above the grid; omitted when empty.
        legend_ax: ``(row, col)`` of the panel carrying the shared legend.

    Returns:
        The completed figure.
    """
    fig, axes = plt.subplots(2, 2, figsize=(11, 8.5), sharey=True)

    panel_labels = "abcd"
    for row in range(2):
        for col in range(2):
            ax = axes[row, col]
            sweeps, title, xlabel, xlim = panels[row][col]
            draw_sweep_panel(ax, sweeps, title, xlabel, xlim)
            label = panel_labels[row * 2 + col]
            ax.text(-0.06, 1.06, f"({label})", transform=ax.transAxes,
                    fontsize=14, fontweight="bold")
        axes[row, 0].set_ylabel(f"{row_labels[row]}\nTest Accuracy (%)")

    axes[0, 0].set_ylim(0, None)
    legend_row, legend_col = legend_ax
    axes[legend_row, legend_col].legend(
        handles=build_legend_handles("SGD-delay", "SGD"),
        loc="upper right", framealpha=0.9, ncol=2)

    if suptitle:
        fig.suptitle(suptitle, y=1.01, fontsize=15)
    fig.tight_layout()
    return fig
```

```python
# ---------- Figure 2: 1st layer / 2nd layer x relocation / jitter (2x2) ----------
fig = draw_2x2_figure(
    [
        [
            (relocation_sweeps, "Random Relocation on 1st Hidden Layer",
             r"Relocation Fraction $f$", (-0.02, 1.02)),
            (jitter_sweeps, "Jitter Perturbation on 1st Hidden Layer",
             r"Jitter Std Dev $\sigma$ (ms)", (-0.5, 26)),
        ],
        [
            (relocation_sweeps_2nd, "Random Relocation on 2nd Hidden Layer",
             r"Relocation Fraction $f$", (-0.02, 1.02)),
            (jitter_sweeps_2nd, "Jitter Perturbation on 2nd Hidden Layer",
             r"Jitter Std Dev $\sigma$ (ms)", (-0.5, 26)),
        ],
    ],
    row_labels=["1st Hidden Layer", "2nd Hidden Layer"],
)
#fig.savefig("combined_1st_2nd_layer_results.png", dpi=150, bbox_inches="tight")
plt.show()
```
